## **Summer School 2026**

TRANSCEND

Las cuencas del Titicaca–Desaguadero–Poopó (TDPS) constituyen un recurso hídrico transfronterizo crítico y vulnerable, compartido por Bolivia y Perú. La agricultura es el principal usuario del agua, representando aproximadamente el 91 % de las extracciones, junto con otros usos como abastecimiento doméstico, minería y energía, además de sostener ecosistemas altoandinos.

Estas cuencas enfrentan importantes presiones ambientales y climáticas, entre ellas la alta variabilidad de las precipitaciones, el aumento de temperaturas y evapotranspiración, el retroceso de glaciares, la expansión agrícola, la sobreexplotación de acuíferos y la reducción del lago Poopó. A ello se suman limitaciones en la disponibilidad y calidad de datos hidrológicos, con una red de monitoreo insuficiente y poca integración de información satelital y agrícola.

El cambio climático, los eventos ENSO, las extracciones no reguladas y las transferencias de agua incrementan la variabilidad y la competencia por el recurso, generando desafíos adicionales para la gestión transfronteriza del agua. Por ello, las instituciones y actores de Bolivia y Perú requieren herramientas que permitan pasar de una gestión reactiva a una gestión preventiva, integrada y basada en evidencia.

En este contexto, la TRANSCEND Summer School 2026 busca fortalecer las capacidades de profesionales e instituciones de los sectores de agua y agricultura mediante la integración de modelación hidrológica, análisis económico de cultivos andinos y monitoreo satelital mediante teledetección, con el objetivo de apoyar una gestión y adaptación más sostenible de los recursos hídricos del TDPS.
