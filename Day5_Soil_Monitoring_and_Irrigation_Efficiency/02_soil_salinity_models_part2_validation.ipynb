```python
# Instalar dependencias
!apt-get update -qq
!apt-get install -y poppler-utils -qq
!pip install gdown pdf2image -q

import gdown
from pdf2image import convert_from_path
from IPython.display import display

# 1. Descargar el PDF (pegar el enlace de Drive)
url = "https://drive.google.com/file/d/1i_AdK54M0g62-pBow6gIKm1Q3TGYZkZA/view?usp=sharing"
gdown.download(url, "documento.pdf", quiet=False, fuzzy=True)

# 2. Convertir páginas a imágenes y mostrarlas
images = convert_from_path("documento.pdf", dpi=150)

print(f"Total de páginas: {len(images)}")
for i, img in enumerate(images):
    print(f"\n--- Página {i+1} ---")
    display(img)
```
